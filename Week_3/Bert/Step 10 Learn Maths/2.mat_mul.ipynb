```python
import torch

# Define the original tensor
K = torch.tensor([[[[ 1.,  2.,  3.],
                    [ 7.,  8.,  9.]],

                   [[ 4.,  5.,  6.],
                    [10., 11., 12.]]],


                  [[[13., 14., 15.],
                    [19., 20., 21.]],

                   [[16., 17., 18.],
                    [22., 23., 24.]]]])

print("Original K Tensor:")
print(K)
print("Shape:", K.shape)  # Output: torch.Size([2, 2, 2, 3])

# Perform transpose on the last two dimensions
K_transposed = K.transpose(-2, -1)

print("\nTransposed K Tensor (transpose(-2, -1)):")
print(K_transposed)
print("Shape:", K_transposed.shape)  # Output: torch.Size([2, 2, 3, 2])
```

```python
# Define the 4x3 matrix with values represented as xij
matrix = [
    ['x11', 'x21'],
    ['y12', 'y22'],
    ['z13', 'z23'],
]

# Transpose the matrix
transposed_matrix = [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]

# Print the transposed matrix
for row in transposed_matrix:
    print(row)
```

```python
# Define the 3x2 matrix with elements labeled as 'aij'
matrix_a = [
    ['x11', 'x21'],
    ['y12', 'y22'],
    ['z13', 'z23']
]

# Define the 2x3 matrix with elements labeled as 'bij'
matrix_b = [
    ['x11', 'y12', 'z13'],
    ['x21', 'y22', 'z23']
]

# Initialize the result matrix (3x3) with empty strings
result = [['' for _ in range(len(matrix_b[0]))] for _ in range(len(matrix_a))]

# Perform matrix multiplication with labels
for i in range(len(matrix_a)):         # iterate over rows of matrix_a
    for j in range(len(matrix_b[0])):   # iterate over columns of matrix_b
        products = []
        for k in range(len(matrix_b)):  # iterate over rows of matrix_b
            products.append(f"({matrix_a[i][k]}*{matrix_b[k][j]})")
        result[i][j] = ' + '.join(products)

# Print the result matrix with symbolic multiplication
for row in result:
    print(row)
```
